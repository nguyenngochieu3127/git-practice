# Visualizing the Titanic Disaster

### Introduction:

This exercise is based on the titanic Disaster dataset avaiable at [Kaggle](https://www.kaggle.com/c/titanic).  
To know more about the variables check [here](https://www.kaggle.com/c/titanic/data)


### Step 1. Import the necessary libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/thieu1995/csv-files/main/data/pandas/titanic_train.csv)

```python
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
```

### Step 3. Assign it to a variable titanic

```python
titanic = pd.read_csv(url)
```

### Step 4. Set PassengerId as the index

```python
titanic.set_index("PassengerId", inplace=True)
```

### Step 5. Create a pie chart presenting the male/female proportion

```python
gender = titanic["Sex"].value_counts()

plt.figure(figsize=(6,6))
plt.pie(
    gender,
    labels=gender.index,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Male/Female Proportion")
plt.show()
```

### Step 6. Create a scatterplot with the Fare payed and the Age, differ the plot color by gender

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=titanic,
    x="Fare",
    y="Age",
    hue="Sex",
    alpha=0.7
)

plt.title("Fare vs Age by Gender")
plt.show()
```

### Step 7. How many people survived?

```python
print("Number of survivors:", titanic["Survived"].sum())
```

### Step 8. Create a histogram with the Fare payed

```python
plt.figure(figsize=(8,6))

sns.histplot(
    titanic["Fare"],
    bins=30,
    kde=True
)

plt.title("Distribution of Fare")
plt.xlabel("Fare")
plt.ylabel("Frequency")
plt.show()
```

### BONUS: Create your own question and answer it.

```python
plt.figure(figsize=(8,6))

sns.boxplot(
    data=titanic,
    x="Pclass",
    y="Fare"
)

plt.title("Fare by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Fare")
plt.show()
```
