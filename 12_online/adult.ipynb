### 20. (6đ) Tỉ lệ người married hay single men (marital-status feature) chiếm nhiều hơn trong số những người kiếm được >50k? Những người có tình trạng hôn nhân (married) có marital-status feature bắt đầu với Married (Married-civ-spouse, Married-spouse-absent or Married-AF-spouse), còn lại được coi là single men.

```python
over_50k = df[df['salary'] == '>50K']

married_patterns = ['Married-civ-spouse', 'Married-spouse-absent', 'Married-AF-spouse']

married_men_over_50k = over_50k[(over_50k['sex'] == 'Male') & (over_50k['marital-status'].isin(married_patterns))].shape[0]
single_men_over_50k = over_50k[(over_50k['sex'] == 'Male') & (~over_50k['marital-status'].isin(married_patterns))].shape[0]

total_men_over_50k = married_men_over_50k + single_men_over_50k

if total_men_over_50k > 0:
    percent_married_men = (married_men_over_50k / total_men_over_50k) * 100
    percent_single_men = (single_men_over_50k / total_men_over_50k) * 100
    print(f"Trong số những người đàn ông kiếm được >50K:")
    print(f"  Tỉ lệ đàn ông đã kết hôn: {percent_married_men:.2f}%")
    print(f"  Tỉ lệ đàn ông độc thân: {percent_single_men:.2f}%")
    if percent_married_men > percent_single_men:
        print("Đàn ông đã kết hôn chiếm tỉ lệ nhiều hơn trong số những người kiếm được >50K.")
    else:
        print("Đàn ông độc thân chiếm tỉ lệ nhiều hơn trong số những người kiếm được >50K.")
else:
    print("Không có đàn ông nào kiếm được >50K trong dữ liệu.")
```

### 21. (6đ) Số giờ tối đa một người làm việc mỗi tuần (hours-per-week feature) là bao nhiêu? Có bao nhiêu người làm việc với số giờ như vậy và tỷ lệ những người kiếm được nhiều (>50K) trong số đó là bao nhiêu?

```python
maxhours = df['hours-per-week'].max()
print(f"Số giờ tối đa một người làm việc mỗi tuần: {maxhours}")

people = df[df['hours-per-week'] == maxhours].shape[0]
print(f"Số người làm việc với số giờ tối đa này: {people}")

over50k = df[(df['hours-per-week'] == maxhours) & (df['salary'] == '>50K')].shape[0]

if people > 0:
    percentover50k = (over50k /people) * 100
    print(f"Tỷ lệ những người kiếm được nhiều (>50K) trong số đó: {percentover50k:.2f}%")
else:
    print("Không có người nào làm việc với số giờ tối đa.")
```
