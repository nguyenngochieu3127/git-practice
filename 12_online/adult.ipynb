### 5. (2đ) Tạo biểu đồ histogram của cột hours-per-week (số giờ làm trong 1 tuần)

```python
# Your code
df['hours-per-week'].hist(bins=20, color = 'c')
plt.title('Distribution of Hours Per Week')
plt.xlabel('Hours Per Week')
plt.ylabel('Frequency')
plt.show()
```

### 6. (3đ) Tạo biểu đồ histogram của cột hours-per-week của những người cố số năm giáo dục (education-num) >= 9.

```python

df[df['education-num'] >= 9]['hours-per-week'].hist(bins=20, color='b',)

plt.title('Phân phối số giờ làm việc/tuần (Education-num >= 9)')
plt.xlabel('Số giờ làm việc mỗi tuần')
plt.ylabel('Tần suất')
plt.show()
```

### 7. (3đ) Tạo pie chart thể hiện phần trăm của cột race và sex (sẽ có 5*2=10 categories tổng thể trong 1 hình vẽ). Nhận xét xem nhóm nào chiếm tỉ lệ nhiều nhất.

```python

race_sex = df.groupby(['race', 'sex']).size()
plt.figure(figsize=(8,8))
plt.pie(race_sex, labels=race_sex.index, autopct='%1.1f%%')
plt.title('Phần trăm dân số theo nhóm race và sex')
plt.show()
print("Nhóm nhiều nhất:", race_sex.idxmax())
```

### 8. (3đ) Tạo pie chart cho cột workclass và nhận xét xem nhóm nào chiếm nhiều nhất.

```python
workclass_counts = df['workclass'].value_counts()
plt.figure(figsize=(8, 8))
plt.pie(workclass_counts, labels=workclass_counts.index, autopct='%1.1f%%', startangle= 130)
plt.title('Tỷ lệ các nhóm Workclass')
plt.show()
print("Nhóm chiếm nhiều nhất là:", workclass_counts.idxmax())
```

### 11. (5đ) Tạo box plot cho cột race và cột education-num (Trục x là các categories của cột race, trục y là giá trị education-num).

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='race', y='education-num', data=df)
plt.title('Education Number by Race')
plt.xlabel('Race')
plt.ylabel('Education Number')
plt.xticks(rotation=45, ha='right')
plt.show()
```

### 12. (5đ) Tạo box plot cho cột workclass và cột hours-per-week (Trực x là các categories của cột workclass, trục y là giá trị hours-per-week).

```python
# Your code
plt.figure(figsize=(12, 6))
sns.boxplot(x='workclass', y='hours-per-week', data=df)
plt.title('Hours Per Week by Workclass')
plt.xlabel('Workclass')
plt.ylabel('Hours Per Week')
plt.xticks(rotation=45, ha='right')
plt.show()
```
